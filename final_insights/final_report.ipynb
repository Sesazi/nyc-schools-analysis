This notebook presents the final summary of NYC school SAT performance and demographic insights.  
It consolidates borough-level results after data cleaning, SQL analysis, and integration.


```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the summary CSV file
data = pd.read_csv('borough_performance.csv')
data
```

```python
plt.figure(figsize=(8, 5))
plt.bar(data['borough'], data['total_schools'], color='skyblue', edgecolor='black')
plt.title('Total Schools per Borough', fontsize=14, weight='bold')
plt.xlabel('Borough')
plt.ylabel('Number of Schools')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.barh(data['borough'], data['avg_sat_math'], color='lightgreen', edgecolor='black')
plt.title('Average SAT Math Scores by Borough', fontsize=14, weight='bold')
plt.xlabel('Average SAT Math Score')
plt.ylabel('Borough')
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.show()
```

## 📊 Final Insights & Recommendations

1️⃣ **Manhattan** leads with the highest SAT Math average (520), showing strong academic performance.  
2️⃣ **Staten Island** follows closely (510), with fewer schools but consistent performance.  
3️⃣ **Bronx** and **Brooklyn** have the most schools but slightly lower SAT averages, suggesting potential overcrowding or resource spread.  
4️⃣ Higher **ELL (English Language Learner)** rates in Bronx and Manhattan highlight linguistic diversity that may require additional support.  

✅ Recommendation:** Focus on equity — invest in targeted programs for language learners and smaller boroughs to balance outcomes.
